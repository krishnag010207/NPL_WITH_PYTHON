# Day 5 Mini Project – SMS Spam Detection

This notebook follows the 9 tasks given in the project instructions using the provided SMS dataset.

## Task 1 – Load and Explore the Dataset

The dataset contains two columns: `Category` and `Message`.

````python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay

# Load the Excel dataset
df = pd.read_excel("spam .xlsx")

# Keep the required columns and clean missing/non-text values
df = df[["Category", "Message"]].copy()
df["Category"] = df["Category"].astype(str).str.strip().str.lower()
df["Message"] = df["Message"].fillna("").astype(str)

print("First 5 rows:")
print(df.head())

print("\nNumber of rows and columns:", df.shape)
print("Columns:", df.columns.tolist())

print("\nNumber of Ham and Spam messages:")
print(df["Category"].value_counts())
````

## Task 2 – Visualize the Dataset

A bar chart is created to compare the number of Spam and Ham messages.

**Answer:** Ham messages are more common in the dataset.

````python
counts = df["Category"].value_counts()

plt.figure(figsize=(7, 5))
plt.bar(counts.index, counts.values)
plt.title("Number of Spam vs Ham Messages")
plt.xlabel("Message Category")
plt.ylabel("Number of Messages")
plt.tight_layout()
plt.show()
````

## Task 3 – Prepare the Data

The target labels are converted into numerical values:

- Ham → 0
- Spam → 1

Then:
- `X` → Message
- `y` → Category

````python
df["Category"] = df["Category"].map({"ham": 0, "spam": 1})

X = df["Message"]
y = df["Category"]

print("X contains SMS messages.")
print("y contains numerical labels (0 = Ham, 1 = Spam).")
````

## Task 4 – Split the Dataset

The dataset is divided into 80% training data and 20% testing data.

**Why split the dataset?**

The training data is used to teach the model, while the testing data is kept separate to check how well the trained model performs on messages it did not see during training.

````python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training messages:", len(X_train))
print("Testing messages:", len(X_test))
````

## Task 5 – Convert Text into Numbers

`CountVectorizer` converts the SMS text into numerical features. Common English stopwords are removed while creating the vectors.

````python
vectorizer = CountVectorizer(stop_words="english")

X_train_vectors = vectorizer.fit_transform(X_train)
X_test_vectors = vectorizer.transform(X_test)

print("Training feature matrix shape:", X_train_vectors.shape)
print("Testing feature matrix shape:", X_test_vectors.shape)
````

## Task 6 – Train a Machine Learning Model

Logistic Regression is trained using the training data. The model learns to distinguish:

- 0 → Ham
- 1 → Spam

````python
model = LogisticRegression(max_iter=1000)
model.fit(X_train_vectors, y_train)

print("Logistic Regression model trained successfully.")
````

## Task 7 – Make Predictions

The trained model predicts the labels of the test messages and stores them in a variable called `predictions`.

````python
predictions = model.predict(X_test_vectors)

print("First 20 predictions:")
print(predictions[:20])
````

## Task 8 – Evaluate the Model

Accuracy and a confusion matrix are calculated. The confusion matrix shows the four required combinations: Actual Ham/Predicted Ham, Actual Ham/Predicted Spam, Actual Spam/Predicted Ham, and Actual Spam/Predicted Spam.

````python
accuracy = accuracy_score(y_test, predictions)
cm = confusion_matrix(y_test, predictions)

print(f"Accuracy: {accuracy:.4f} ({accuracy * 100:.2f}%)")
print("\nConfusion Matrix:")
print(cm)

correct = int((predictions == y_test).sum())
print(f"\nCorrectly classified messages: {correct}")
print(f"Total test messages: {len(y_test)}")

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Ham", "Spam"]
).plot()
plt.title("SMS Spam Detection – Confusion Matrix")
plt.tight_layout()
plt.show()
````

### Evaluation result from the provided dataset

- Accuracy: **97.49%**
- Correctly classified test messages: **1,087 out of 1,115**
- Confusion matrix:

```text
[[965   1]
 [ 27 122]]
```

## Task 9 – Test Your Own SMS

The user enters a new SMS using `input()`. The same CountVectorizer used during training transforms the new message before the trained model makes its prediction.

````python
new_message = input("Enter an SMS message: ")

new_message_vector = vectorizer.transform([new_message])
prediction = model.predict(new_message_vector)[0]

if prediction == 0:
    print("📩 HAM MESSAGE")
else:
    print("🚨 SPAM MESSAGE")
````
